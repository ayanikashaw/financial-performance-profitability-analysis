```python
import os

base_path = "financial_data/Detailed-Financials-Data-Of-4456-NSE-And-BSE-Company-20231230T233228Z-001/Detailed-Financials-Data-Of-4456-NSE-_-BSE-Company"

company = "Tata Consumer Products Ltd"

os.listdir(os.path.join(base_path, company))
```

```python
import pandas as pd
file_path = os.path.join(
    base_path,
    "Tata Consumer Products Ltd",
    "Yearly_Profit_Loss.csv"
)
profit_loss = pd.read_csv(file_path)
profit_loss.head()
```

```python
profit_loss = pd.read_csv(file_path, index_col=0)
profit_loss
```

```python
profit_loss = profit_loss.T
profit_loss
```

```python
# Calculate profitability metrics
profit_loss["Net Profit Margin %"] = (
    profit_loss["Net Profit"] / profit_loss["Sales"]
) * 100
profit_loss["Expense Ratio %"] = (
    profit_loss["Expenses"] / profit_loss["Sales"]
) * 100
profit_loss
```

```python
# Calculate year-over-year growth
profit_loss["Sales Growth %"] = profit_loss["Sales"].pct_change() * 100
profit_loss["Net Profit Growth %"] = profit_loss["Net Profit"].pct_change() * 100
profit_loss
```

```python
# Plot Sales and Net Profit trend

import matplotlib.pyplot as plt
plt.figure(figsize=(10, 5))
plt.plot(profit_loss.index[:-1], profit_loss["Sales"][:-1], marker="o", label="Sales")
plt.plot(profit_loss.index[:-1], profit_loss["Net Profit"][:-1], marker="o", label="Net Profit")

plt.title("Sales and Net Profit Trend")
plt.xlabel("Year")
plt.ylabel("Amount (₹ Crore)")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Plot Net Profit Margin and Operating Profit Margin

plt.figure(figsize=(10, 5))

plt.plot(
    profit_loss.index[:-1],
    profit_loss["Net Profit Margin %"][:-1],
    marker="o",
    label="Net Profit Margin"
)

plt.plot(
    profit_loss.index[:-1],
    profit_loss["OPM %"][:-1],
    marker="o",
    label="Operating Profit Margin"
)

plt.title("Profitability Margin Trend")
plt.xlabel("Year")
plt.ylabel("Margin (%)")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Plot Expense Ratio

plt.figure(figsize=(10, 5))

plt.plot(
    profit_loss.index[:-1],
    profit_loss["Expense Ratio %"][:-1],
    marker="o"
)

plt.title("Expense Ratio Trend")
plt.xlabel("Year")
plt.ylabel("Expense Ratio (%)")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
# Plot EPS trend

plt.figure(figsize=(10, 5))
plt.plot(
    profit_loss.index[:-1],
    profit_loss["EPS in Rs"][:-1],
    marker="o"
)
plt.title("EPS Trend")
plt.xlabel("Year")
plt.ylabel("EPS (₹)")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
# Create a summary of key financial metrics

summary = profit_loss.loc[
    ["Mar 2012", "Mar 2023", "TTM"],
    ["Sales", "Operating Profit", "Net Profit", "OPM %",
     "Net Profit Margin %", "Expense Ratio %", "EPS in Rs"]
]
summary
```

```python
# Find the best and worst years based on Net Profit

best_year = profit_loss["Net Profit"].idxmax()
worst_year = profit_loss["Net Profit"].idxmin()

print("Best year:", best_year)
print("Highest Net Profit:", profit_loss.loc[best_year, "Net Profit"])

print("\nWorst year:", worst_year)
print("Lowest Net Profit:", profit_loss.loc[worst_year, "Net Profit"])
```

```python
# Calculate overall growth from 2012 to 2023

sales_growth = (
    (profit_loss.loc["Mar 2023", "Sales"] /
     profit_loss.loc["Mar 2012", "Sales"]) - 1
) * 100

profit_growth = (
    (profit_loss.loc["Mar 2023", "Net Profit"] /
     profit_loss.loc["Mar 2012", "Net Profit"]) - 1
) * 100

print("Sales growth from 2012 to 2023:", round(sales_growth, 2), "%")
print("Net profit growth from 2012 to 2023:", round(profit_growth, 2), "%")
```

```python
# Calculate CAGR from 2012 to 2023

years = 2023 - 2012
sales_cagr = (
    (profit_loss.loc["Mar 2023", "Sales"] /
     profit_loss.loc["Mar 2012", "Sales"]) ** (1 / years) - 1
) * 100
profit_cagr = (
    (profit_loss.loc["Mar 2023", "Net Profit"] /
     profit_loss.loc["Mar 2012", "Net Profit"]) ** (1 / years) - 1
) * 100
print("Sales CAGR:", round(sales_cagr, 2), "%")
print("Net Profit CAGR:", round(profit_cagr, 2), "%")
```

```python
# Load yearly cash flow data

cash_flow_path = os.path.join(
    base_path,
    "Tata Consumer Products Ltd",
    "Yearly_Cash_flow.csv"
)

cash_flow = pd.read_csv(cash_flow_path, index_col=0)

cash_flow
```

```python
# Arrange cash flow data year-wise

cash_flow = cash_flow.T

cash_flow
```

```python
# Plot operating cash flow and net cash flow

plt.figure(figsize=(10, 5))

plt.plot(
    cash_flow.index,
    cash_flow["Cash from Operating Activity"],
    marker="o",
    label="Operating Cash Flow"
)

plt.plot(
    cash_flow.index,
    cash_flow["Net Cash Flow"],
    marker="o",
    label="Net Cash Flow"
)

plt.title("Cash Flow Trend")
plt.xlabel("Year")
plt.ylabel("Amount (₹ Crore)")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Key cash flow observations

highest_operating_cash = cash_flow["Cash from Operating Activity"].idxmax()
lowest_net_cash = cash_flow["Net Cash Flow"].idxmin()

print("Highest Operating Cash Flow:", highest_operating_cash)
print(
    "Amount:",
    cash_flow.loc[highest_operating_cash, "Cash from Operating Activity"],
    "₹ Crore"
)

print("\nLowest Net Cash Flow:", lowest_net_cash)
print(
    "Amount:",
    cash_flow.loc[lowest_net_cash, "Net Cash Flow"],
    "₹ Crore"
)
```
